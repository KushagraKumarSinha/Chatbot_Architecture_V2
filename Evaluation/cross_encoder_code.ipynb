```python
from sentence_transformers import CrossEncoder
import pandas as pd

# Initialize the Cross-Encoder model
# 'cross-encoder/ms-marco-MiniLM-L-6-v2' is commonly used for search/retrieval relevancy.
# 'cross-encoder/stsb-distilroberta-base' is another great alternative for general semantic similarity.
model_name = 'cross-encoder/ms-marco-MiniLM-L-6-v2'
model = CrossEncoder(model_name)

# Define the query and retrieved context pairs you want to evaluate
query = "What is the capital of France?"
retrieved_contexts = [
    "Paris is the capital and most populous city of France.",
    "The Eiffel Tower is a famous landmark located in Paris, France.",
    "Berlin is the capital of Germany and its largest city by both area and population."
]

# Format as a list of pairs: [ [query, doc1], [query, doc2], ... ]
pairs = [[query, doc] for doc in retrieved_contexts]

# Predict similarity scores
scores = model.predict(pairs)

# Display the results
results = pd.DataFrame({
    'Query': [query] * len(retrieved_contexts),
    'Retrieved Context': retrieved_contexts,
    'Similarity Score': scores
}).sort_values(by='Similarity Score', ascending=False)

display(results)
```
